Compute and report the overall performance of Poly for orders 1:8 for any one of the other features in the Auto dataset (acceleration) to mpg

```r
library(boot)

loocv <- rep(0, 8)
for (i in 1:8) {
  loocv.fit <- glm(mpg ~ poly(acceleration, i), data = Auto)
  loocv[i] <- cv.glm(Auto, loocv.fit)$delta[1]
}
loocv
```

Select a different feature form Auto than you did in task 3 and perform a 5 fold and 10 fold k-fold cross validation

```r
set.seed(0397)
cv.error.5 <- rep(0, 10)
for (i in 1:10) {
  glm.fit <- glm(mpg ~ weight, data = Auto)
  cv.error.5[i] <- cv.glm(Auto, glm.fit, K = 5)$delta[1]
}
cv.error.5
```

Compute the last bootstrap exercise (the quadratic fit for horsepower) from the lab 3 more times, using a new number of samples {250, 500, 2500}

```r
boot.fn <- function(data, index)
  coef(
      lm(mpg ~ horsepower + I(horsepower^2),
        data = data, subset = index)
    )
set.seed(0397)
boot(Auto, boot.fn, 250)
```
