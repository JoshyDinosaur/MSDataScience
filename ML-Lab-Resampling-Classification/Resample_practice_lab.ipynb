```r
library(ISLR2)
set.seed(0397)
train <- sample(392, 196)

attach(Auto)
```

```r
lm.fit2 <- lm(mpg ~ poly(horsepower, 2), data = Auto,
    subset = train)
mean((mpg - predict(lm.fit2, Auto))[-train]^2)
```

```r
lm.fit3 <- lm(mpg ~ poly(horsepower, 3), data = Auto,
    subset = train)
mean((mpg - predict(lm.fit3, Auto))[-train]^2)
```

```r
library(boot)

loocv <- rep(0, 8)
for (i in 1:8) {
  loocv.fit <- glm(mpg ~ poly(acceleration, i), data = Auto)
  loocv[i] <- cv.glm(Auto, loocv.fit)$delta[1]
}
loocv
```

```r

cv.error.5 <- rep(0, 8)
for (i in 1:8) {
  glm.fit <- glm(mpg ~ poly(weight,i), data = Auto)
  set.seed(0397)
  cv.error.5[i] <- cv.glm(Auto, glm.fit, K = 5)$delta[1]
  set.seed(0397)
}
cv.error.5
```

```r

cv.error.10 <- rep(0, 8)
for (i in 1:8) {
  glm.fit <- glm(mpg ~ displacement, data = Auto)
  set.seed(0397)
  cv.error.10[i] <- cv.glm(Auto, glm.fit, K = 10)$delta[1]
  set.seed(0397)
}
cv.error.10
```

```r
boot.fn <- function(data, index)
  coef(
      lm(mpg ~ horsepower + I(horsepower^2),
        data = data, subset = index)
    )
set.seed(1)
boot(Auto, boot.fn, 250)
```
