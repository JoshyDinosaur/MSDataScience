## Task 6 — Build a cleaned copy (12 points)

Create <code>df_clean</code> as a copy of <code>df_raw</code>. Do not change <code>df_raw</code>.

Apply all of these rules to <code>df_clean</code>:

1. Fill missing numeric values with each column's median.
2. Fill missing non-numeric values with each column's mode.
3. Remove duplicate rows.
4. Reset the index after duplicate removal.

Finally, display:

- total missing values before and after cleaning;
- duplicate row count after cleaning;
- the shape of the cleaned DataFrame.

```python
# Task 6 solution
df_clean = df_raw.copy()

numeric_columns = df_clean.select_dtypes(include="number").columns
categorical_columns = df_clean.select_dtypes(exclude="number").columns

numeric_fill_values = df_clean[numeric_columns].median()
categorical_fill_values = df_clean[categorical_columns].mode().iloc[0]

df_clean[numeric_columns] = df_clean[numeric_columns].fillna(numeric_fill_values)
df_clean[categorical_columns] = df_clean[categorical_columns].fillna(
    categorical_fill_values
)

missing_values = int(df_raw.isna().sum().sum())

duplicate_count = df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates()

df_clean = df_clean.reset_index(drop=True)

missing_values_after = int(df_clean.isna().sum().sum())

print("Missing values:", missing_values)
print("Missing values after: ", missing_values_after)
print(f"Duplicate rows removed: {duplicate_count}")
print(f"Cleaned DataFrame shape: {df_clean.shape}")
```
